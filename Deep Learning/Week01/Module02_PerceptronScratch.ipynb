# Implementationof perceptron from scratch

```python
import numpy as np
import matplotlib.pyplot as plt
import joblib
import pandas as pd
plt.style.use("fivethirtyeight")
```

```python
class Perceptron:
  def __init__(self, eta, epochs):
    self.weights = np.random.randn(3)* 1e-4
    self.eta = eta
    self.epochs = epochs

  def activationFunction(self, inputs, weights):
    z = np.dot(inputs, weights)
    return np.where(z>0, 1, 0)

  def fit(self, X, y):
    self.X = X
    self.y = y
    X_with_bias = np.c_[self.X, -np.ones((len(self.X), 1))]
    print(f"X with bias: {X_with_bias}/n")

    for epoch in range(self.epochs):
      print("--"*10)
      print(f"for epoch >> {epoch}")
      print("--"*10)

      y_hat = self.activationFunction(X_with_bias, self.weights)
      print(f"predicted value after forward pass: {y_hat}")

      error = self.y - y_hat
      print(f"error: {error}")

      self.weights = self.weights + self.eta * np.dot(X_with_bias.T, error) # weight updatation
      print(f"updated weights after epoch: {epoch}/{self.epochs}: {self.weights}")


  def predict(self, X):
    X_with_bias = np.c_[self.X, -np.ones((len(self.X), 1))]
    return self.activationFunction(X_with_bias, self.weights) # prediction will come from the activation function, as weights has been updated in fit()
```

```python
data = {"x1": [0,0,1,1], "x2": [0,1,0,1], "y": [0,0,0,1]}

AND = pd.DataFrame(data)
AND
```

```python
X = AND.drop("y", axis=1)
y = AND['y']
y.to_frame()
```

```python
model = Perceptron(eta = 0.1, epochs = 10)
model.fit(X,y)
```

```python
model.predict(X)
```

# Saving model

```python
import os
dir_ = "Perceptron_model"
os.makedirs(dir_, exist_ok=True)
filename = os.path.join(dir_, "AND.model")
joblib.dump(model, filename)
```

```python
# load the model from drive:

loaded_model = joblib.load(filename)
result = loaded_model.predict(X)
result
```
